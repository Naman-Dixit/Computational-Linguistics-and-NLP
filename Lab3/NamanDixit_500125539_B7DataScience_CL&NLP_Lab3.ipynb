```python
# ============================================================
# CELL 2: IMPORT LIBRARIES
# ============================================================

import requests
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.stem import PorterStemmer

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

print("All libraries loaded successfully!")
```

```python
# ============================================================
# CELL 3: DOWNLOAD COMMUNISM ARTICLE
# ============================================================

# Wikipedia API URL
api_url = "https://en.wikipedia.org/w/api.php"

# Parameters for the API
params = {
    "action": "query",
    "format": "json",
    "prop": "extracts",
    "explaintext": True,
    "titles": "Communism",
    "redirects": True
}

# Send request to Wikipedia
response = requests.get(
    api_url,
    params=params,
    headers={
        "User-Agent": "NLP-Lab-Student/1.0"
    }
)

# Check whether request was successful
print("Status Code:", response.status_code)

if response.status_code == 200:
    print("Wikipedia article downloaded successfully!")
else:
    print("Error downloading article.")
```

```python
# ============================================================
# CELL 4: EXTRACT ARTICLE TEXT
# ============================================================

# Convert response into JSON
data = response.json()

# Get the pages dictionary
pages = data["query"]["pages"]

# Get the first page
page = next(iter(pages.values()))

# Extract plain text
raw_text = page["extract"]

print("Article extracted successfully!")

print("Number of characters:", len(raw_text))

print("\n")
print("============================================================")
print("BEGINNING OF COMMUNISM ARTICLE")
print("============================================================")

# Display first 5000 characters
print(raw_text[:5000])
```

```python
# ============================================================
# CELL 5: CREATE corpus.txt
# ============================================================

# Save the Wikipedia article into a text file

with open("corpus.txt", "w", encoding="utf-8") as file:
    file.write(raw_text)

print("corpus.txt created successfully!")
```

```python
# ============================================================
# CELL 6: READ corpus.txt
# ============================================================

with open("corpus.txt", "r", encoding="utf-8") as file:
    corpus = file.read()

print("Corpus loaded successfully!")

print("Number of characters:", len(corpus))

print("\nFirst 2000 characters:")
print(corpus[:2000])
```

```python
# ============================================================
# CELL 7: TOKENIZATION
# ============================================================

# Convert the article into individual tokens
tokens = word_tokenize(corpus)

print("Total number of tokens:", len(tokens))

print("\nFirst 100 tokens:")
print(tokens[:100])
```

```python
# ============================================================
# CELL 8: STOP-WORD REMOVAL
# ============================================================

# Get English stop words
stop_words = set(stopwords.words("english"))

# Create an empty list
cleaned_tokens = []

# Go through every token
for word in tokens:

    # Convert to lowercase
    word = word.lower()

    # Keep only alphabetic words
    # Remove stop words
    if word.isalpha() and word not in stop_words:
        cleaned_tokens.append(word)

print("Tokens before cleaning:", len(tokens))
print("Tokens after cleaning:", len(cleaned_tokens))

print("\nFirst 100 cleaned tokens:")
print(cleaned_tokens[:100])
```

```python
# ============================================================
# CELL 9: LEMMATIZATION
# ============================================================

# Create WordNet lemmatizer
lemmatizer = WordNetLemmatizer()

# Empty list for lemmatized words
lemmatized_tokens = []

# Lemmatize every cleaned word
for word in cleaned_tokens:

    lemma = lemmatizer.lemmatize(word)

    lemmatized_tokens.append(lemma)

print("First 100 lemmatized tokens:")
print(lemmatized_tokens[:100])
```

```python
# ============================================================
# CELL 10: SAVE cleaned_corpus.txt
# ============================================================

# Join all words into one string
cleaned_text = " ".join(lemmatized_tokens)

# Save the cleaned text
with open("cleaned_corpus.txt", "w", encoding="utf-8") as file:
    file.write(cleaned_text)

print("cleaned_corpus.txt created successfully!")
```

```python
# ============================================================
# CELL 11: CHECK CLEANED CORPUS
# ============================================================

with open("cleaned_corpus.txt", "r", encoding="utf-8") as file:
    cleaned_corpus = file.read()

print("============================================================")
print("CLEANED CORPUS")
print("============================================================")

print(cleaned_corpus[:3000])
```

```python
# ============================================================
# CELL 12: SPACY PRE-PROCESSING
# ============================================================

# Process the entire Wikipedia article using spaCy
doc = nlp(corpus)

# Store processed words
spacy_processed = []

for token in doc:

    # Remove stop words
    # Remove punctuation
    # Keep only alphabetic words
    if not token.is_stop and token.is_alpha:

        # Convert to lowercase and get lemma
        spacy_processed.append(token.lemma_.lower())

print("First 100 spaCy processed tokens:")
print(spacy_processed[:100])

print("\nTotal processed tokens:", len(spacy_processed))
```

```python
# ============================================================
# CELL 18: Excercise 3 - MORPHOLOGICAL FEATURE EXTRACTION
# ============================================================

def extract_morphological_features(text):

    # Process text using spaCy
    doc = nlp(text)

    # Store results
    result = []

    # Examine every token
    for token in doc:

        # Ignore punctuation and numbers
        if not token.is_alpha:
            continue

        # Get morphological information
        morphology = str(token.morph)

        # ----------------------------------------------------
        # CHECK FOR PLURAL NOUN
        # ----------------------------------------------------

        if token.pos_ == "NOUN" and "Number=Plur" in morphology:

            result.append(
                (
                    token.text,
                    token.lemma_,
                    "Plural Noun",
                    morphology
                )
            )

        # ----------------------------------------------------
        # CHECK FOR PAST TENSE VERB
        # ----------------------------------------------------

        elif token.pos_ == "VERB" and "Tense=Past" in morphology:

            result.append(
                (
                    token.text,
                    token.lemma_,
                    "Past Tense Verb",
                    morphology
                )
            )

    return result
```

```python
# ============================================================
# CELL 19: RUN MORPHOLOGICAL ANALYSIS
# ============================================================

features = extract_morphological_features(corpus)

print("Total morphological features found:")
print(len(features))

print("\n")
print("First 100 results:")
print("=" * 80)

for word, lemma, feature, morphology in features[:100]:

    print(
        f"Word: {word:<20} "
        f"Lemma: {lemma:<20} "
        f"Type: {feature:<20} "
        f"Morphology: {morphology}"
    )
```

```python
# ============================================================
# CELL 20: SEPARATE FEATURES
# ============================================================

plural_nouns = []
past_tense_verbs = []

# Go through all extracted features
for word, lemma, feature, morphology in features:

    if feature == "Plural Noun":

        plural_nouns.append(word)

    elif feature == "Past Tense Verb":

        past_tense_verbs.append(word)


# Remove duplicates
unique_plural_nouns = sorted(set(plural_nouns))

unique_past_tense_verbs = sorted(set(past_tense_verbs))


print("============================================================")
print("PLURAL NOUNS")
print("============================================================")

print(unique_plural_nouns[:100])


print("\n")
print("============================================================")
print("PAST TENSE VERBS")
print("============================================================")

print(unique_past_tense_verbs[:100])
```

```python
# ============================================================
# CELL 21: COUNT MORPHOLOGICAL FEATURES
# ============================================================

print("Morphological Analysis Summary")
print("=" * 50)

print(
    "Total plural noun occurrences:",
    len(plural_nouns)
)

print(
    "Unique plural nouns:",
    len(unique_plural_nouns)
)

print(
    "Total past tense verb occurrences:",
    len(past_tense_verbs)
)

print(
    "Unique past tense verbs:",
    len(unique_past_tense_verbs)
)
```
