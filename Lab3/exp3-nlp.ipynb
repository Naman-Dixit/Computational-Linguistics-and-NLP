### Text Pre-processing & Morphological Analysis

#### 1. Basic Text Pre-processing
Raw text data is often noisy and unstructured. Pre-processing transforms it into a predictable format for machine learning models:
*   **Tokenization:** Segmenting text into smaller units (tokens), typically words or sentences.
*   **Stop-word Removal:** Eliminating highly common but semantically empty words (e.g., "is", "the", "and") to reduce the dimensionality of the data.
*   **Lemmatization:** Reducing words to their base or dictionary form (lemma). Unlike stemming, which just chops off affixes, lemmatization considers the context and part of speech (e.g., "better" becomes "good", "running" becomes "run").

#### 2. Morphological Analysis
Morphology is the study of the internal structure of words. Morphological analysis involves identifying a word's root/stem and its affixes (prefixes/suffixes) to understand its grammatical role (tense, number, gender, case).
*   **Stemming (Heuristic):** A crude heuristic process that chops off the ends of words (e.g., "happiness" $\rightarrow$ "happi").
*   **Deep Morphological Features:** Identifying structural grammar attributes like `Tense=Past` or `Number=Plur`.

# Basic Pre-processing (Tokenization, Stop-words, Lemmatization)

```python
# Reading a document and applying NLTK Pre-processing

raw_text = """
The bats were hanging upside down in the caves. 
The children are playing outside and running very fast! 
He has successfully completed his multiple assignments.
"""

print(f"--- Original Text ---\n{raw_text.strip()}\n")

# 1. Tokenization
tokens = word_tokenize(raw_text)

# 2. Stop-word Removal (and lowercase conversion + punctuation removal)
stop_words = set(stopwords.words('english'))
cleaned_tokens = [
    word.lower() for word in tokens 
    if word.isalpha() and word.lower() not in stop_words
]

# 3. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in cleaned_tokens]

print("--- NLTK Pre-processing Pipeline ---")
print(f"1. Raw Tokens: {tokens[:10]}...")
print(f"2. Cleaned Tokens (No Punct/Stopwords): {cleaned_tokens}")
print(f"3. Lemmatized Tokens: {lemmatized_tokens}")
```

```python
# Pre-processing using spaCy (Alternative, streamlined approach)

# Load spaCy English pipeline
nlp = spacy.load("en_core_web_sm")

# Process the text
doc = nlp(raw_text)

# spaCy does tokenization, lemmatization, and stop-word identification in one step
spacy_processed = [
    token.lemma_.lower() for token in doc 
    if not token.is_stop and token.is_alpha
]

print("--- spaCy Pre-processing Pipeline ---")
print(f"Lemmatized & Cleaned Tokens: {spacy_processed}")
```

# Study and Implementation of Morphological Analysis

```python
# Morphological Analysis using NLTK (Stemming)

stemmer = PorterStemmer()
test_words = ["running", "runs", "ran", "happiness", "unbelievable", "studies"]

print("--- NLTK Morphological Analysis (Stemming) ---")
print(f"{'Original Word':<15} | {'Stemmed Root':<15}")
print("-" * 35)
for word in test_words:
    print(f"{word:<15} | {stemmer.stem(word):<15}")
```

```python
# Deep Morphological Analysis using spaCy

morph_text = "The developers are actively upgrading the server databases."
morph_doc = nlp(morph_text)

print("--- spaCy Detailed Morphological Analysis ---")
print(f"{'Token':<12} | {'Lemma':<10} | {'POS Tag':<8} | {'Morphological Features'}")
print("-" * 75)

for token in morph_doc:
    if token.is_alpha:
        print(f"{token.text:<12} | {token.lemma_:<10} | {token.pos_:<8} | {token.morph}")
```
